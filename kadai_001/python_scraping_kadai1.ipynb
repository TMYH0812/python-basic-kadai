```python
import requests
from bs4 import BeautifulSoup

# URLからHTMLを取得
url="https://news.yahoo.co.jp/articles/43dbf22f25d14012f3134732f3511dc58d2e41ef"
response = requests.get(url)
soup = BeautifulSoup(response.text,'html.parser')

# すべての<p>タグを取得
sentences = soup.select('#uamods .article_body p.highLightSearchTarget')

# 各<p>タグのテキストを結合して1つの文字列にする
article_text = ''.join([sentence.text for sentence in sentences])

print(article_text)
```
