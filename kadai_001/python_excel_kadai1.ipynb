```python
import openpyxl
from openpyxl.styles import Border, Font, Side
from datetime import datetime


# 新しいExcelブックとワークシートを作成
workbook = openpyxl.Workbook()
sheet = workbook.active
sheet.title = '請求書'

# 請求書のタイトル
sheet['B2'] = '請求書'
sheet['B2'].font = Font(size=14, bold=True)

# 会社情報
sheet['B4'] = '株式会社ABC'
sheet['B5'] = '〒101-0022 東京都千代田区神田練塀町300'
sheet['B6'] = 'TEL:03-1234-5678 FAX:03-1234-5678'
sheet['B7'] = '担当者名:鈴木一郎 様'

# 請求番号と日付
sheet['F4'] = 'No.'
sheet['G4'] = '0001'
sheet['F5'] = '日付'
sheet['G5'] = datetime.now()
sheet['G5'].number_format = 'yyyy/mm/dd'

# 商品明細の見出し
header = ['商品名', '数量', '単価', '金額']
for column, value in enumerate(header, start=2):
    sheet.cell(row=10, column=column, value=value)

# 商品データ
data = [
    ['商品A', 2, 10000],
    ['商品B', 1, 15000]
]

for row_number, row_data in enumerate(data, start=11):
    sheet.cell(row=row_number, column=2, value=row_data[0])
    sheet.cell(row=row_number, column=3, value=row_data[1])
    sheet.cell(row=row_number, column=4, value=row_data[2])
    sheet.cell(row=row_number, column=5, value=f'=C{row_number}*D{row_number}')

# 明細金額の合計
sheet['E13'] = '=SUM(E11:E12)'

# 小計・消費税・合計
sheet['B15'] = '小計'
sheet['E15'] = '=E13'
sheet['B16'] = '消費税'
sheet['E16'] = '=E15*10%'
sheet['B17'] = '合計'
sheet['E17'] = '=E15+E16'

# 見出しと合計欄を太字に設定
for cell in ['B10', 'C10', 'D10', 'E10', 'B15', 'B16', 'B17']:
    sheet[cell].font = Font(bold=True)

# 表に罫線を設定
thin_border = Border(
    left=Side(style='thin'),
    right=Side(style='thin'),
    top=Side(style='thin'),
    bottom=Side(style='thin')
)

for row in sheet['B10:E12']:
    for cell in row:
        cell.border = thin_border

# 数値の表示形式を設定
for row_number in range(11, 18):
    for column in range(3, 6):
        sheet.cell(row=row_number, column=column).number_format = '#,##0'

# 列幅を調整
sheet.column_dimensions['A'].width = 2
sheet.column_dimensions['B'].width = 18
sheet.column_dimensions['C'].width = 10
sheet.column_dimensions['D'].width = 12
sheet.column_dimensions['E'].width = 14
sheet.column_dimensions['F'].width = 10
sheet.column_dimensions['G'].width = 14

# 実行日を付けたファイル名で保存
today = datetime.now().strftime('%Y%m%d')
workbook.save(f'請求書_{today}.xlsx')
```
