```python
import requests
from getpass import getpass

# APIキーを指定
api_key = getpass('APIキーを入力してください: ').strip()

# 検索したい駅名を入力
station_name = input("駅名を入力してください：")

# 検索キーワードを作成
query = f"{station_name}駅 レストラン"

# Places APIのURL
url = "https://maps.googleapis.com/maps/api/place/textsearch/json"

# APIへ渡す情報
params = {
    "query": query,
    "language": "ja",
    "key": api_key
}

try:
    # GETメソッドでデータを取得
    response = requests.get(url, params=params)
    response.raise_for_status() # HTTPエラーが発生した場合に例外を発生させる

    # JSON形式に変換
    data = response.json()

    if data.get("status") == "OK":
        restaurants = data.get("results", [])

        for restaurant in restaurants:
            name = restaurant.get("name", "店舗名なし")
            address = restaurant.get("formatted_address", "住所なし")
            rating = restaurant.get("rating", "評価なし")

            print(f"店舗名：{name}")
            print(f"住所：{address}")
            print(f"評価：{rating}")
            print("-" * 30)

    else:
        print("データを取得できませんでした。")
        print("ステータス：", data.get("status"))
        # APIからのエラーメッセージがあれば表示
        print("エラー内容：", data.get("error_message", "不明なエラー"))

except requests.exceptions.RequestException as error:
    print(f"APIへの接続に失敗しました、またはHTTPエラーが発生しました: {error}")
except ValueError:
    print("APIの応答をJSONとして解析できませんでした。レスポンスが不正なJSON形式かもしれません。")
except Exception as e:
    print(f"予期せぬエラーが発生しました: {e}")
```
