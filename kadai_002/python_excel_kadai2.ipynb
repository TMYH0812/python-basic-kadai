```python
import pandas as pd

# データフレームの作成
df = pd.DataFrame({
    '日付':
      ['2023-05-17', '2023-05-18', '2023-05-19', '2023-05-20', '2023-05-21'],
    '社員名': ['山田', '佐藤', '鈴木', '田中', '高橋'],
    '売上': [100, 200, 150, 300, 250],
    '部門': ['メーカー', '代理店', 'メーカー', '商社', '代理店'],
})

# 全体の売上の平均を求め、「平均売上」列を作成
average_sales = df['売上'].mean()
df['平均売上'] = average_sales

# 売上を平均売上と比較して業績ランクを返す関数
def performance(sales):
    if sales >= average_sales + 50:
        return 'A'
    elif sales >= average_sales:
        return 'B'
    else:
        return 'C'

# 「業績ランク」列を作成
df['業績ランク'] = df['売上'].apply(performance)

# 行番号を含めず、新規Excelファイルに書き出す
df.to_excel('業績.xlsx', sheet_name='Sheet1', index=False)
```
