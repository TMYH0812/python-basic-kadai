```python
import pandas as pd
from openpyxl.styles import PatternFill

# Sheet1をデータフレームとして読み込む
df_2022 = pd.read_excel("2022_年間売上表.xlsx", sheet_name="Sheet1")
df_2023 = pd.read_excel("2023_年間売上表.xlsx", sheet_name="Sheet1")

# 2年分のデータを縦方向に連結する
df_combined = pd.concat([df_2022, df_2023], ignore_index=True)

# 「商品」「売上年」ごとに金額を合計する
df_summary = (
    df_combined
    .groupby(["商品", "売上年"], as_index=False)["金額（千円）"]
    .sum()
)

# 新規Excelファイルに出力し、ヘッダーを薄いグレーにする
with pd.ExcelWriter("売上集計表.xlsx", engine="openpyxl") as writer:
    df_summary.to_excel(writer, sheet_name="Sheet1", index=False)

    worksheet = writer.sheets["Sheet1"]
    header_fill = PatternFill(
        fill_type="solid",
        fgColor="F2F2F2"
    )

    for cell in worksheet[1]:
        cell.fill = header_fill

print("売上集計表.xlsxを保存しました。")
```
